```python
import numpy

class DataAnalytics:
    def __init__(self):
        """
        Constructor to start array.
        Arguments taken: self.
        Return value: none.
        """
        self.data = None
    def __check(self):
        """
        Private method to check data.
        Arguments taken: self.
        Return value: boolean.
        """
        if self.data is None:
            print("Make array first")
            return True
        return False
    def show(cls):
        """
        Class method to print name.
        Arguments taken: cls.
        Return value: none.
        """
        print("DataAnalytics class is running")
    def create(self):
        """
        Method to create numpy array.
        Arguments taken: self.
        Return value: none.
        """
        print("1 for 1D")
        print("2 for 2D")
        print("3 for 3D")
        choice = input("Enter choice: ")
        
        if choice == "1":
            text = input("Enter numbers separated by space: ")
            words = text.split()
            numbers = []
            for word in words:
                numbers.append(int(word))
            self.data = numpy.array(numbers)
            print("Array created")
            print(self.data)
            
            print("1 for Indexing")
            print("2 for Slicing")
            action = input("Enter choice: ")
            if action == "1":
                index = int(input("Enter index: "))
                print(self.data[index])
            if action == "2":
                start = int(input("Enter start: "))
                end = int(input("Enter end: "))
                print(self.data[start:end])
                
        if choice == "2":
            rows = int(input("Enter rows: "))
            cols = int(input("Enter columns: "))
            text = input("Enter numbers separated by space: ")
            words = text.split()
            numbers = []
            for word in words:
                numbers.append(int(word))
            self.data = numpy.array(numbers).reshape(rows, cols)
            print("Array created")
            print(self.data)
            
        if choice == "3":
            blocks = int(input("Enter blocks: "))
            rows = int(input("Enter rows: "))
            cols = int(input("Enter columns: "))
            text = input("Enter numbers separated by space: ")
            words = text.split()
            numbers = []
            for word in words:
                numbers.append(int(word))
            self.data = numpy.array(numbers).reshape(blocks, rows, cols)
            print("Array created")
            print(self.data)

    def maths(self):
        """
        Method to do maths.
        Arguments taken: self.
        Return value: none.
        """
        empty = self.__check()
        if empty == True:
            return
            
        print("1 for Add")
        print("2 for Subtract")
        print("3 for Multiply")
        print("4 for Divide")
        print("5 for Dot Product")
        choice = input("Enter choice: ")
        
        text = input("Enter second array numbers: ")
        words = text.split()
        numbers = []
        for word in words:
            numbers.append(int(word))
        second = numpy.array(numbers).reshape(self.data.shape)
        
        if choice == "1":
            answer = numpy.add(self.data, second)
            print(answer)
        if choice == "2":
            answer = numpy.subtract(self.data, second)
            print(answer)
        if choice == "3":
            answer = numpy.multiply(self.data, second)
            print(answer)
        if choice == "4":
            answer = numpy.divide(self.data, second)
            print(answer)
        if choice == "5":
            answer = numpy.dot(self.data, second)
            print(answer)

    def split_combine(self):
        """
        Method to combine and split.
        Arguments taken: self.
        Return value: none.
        """
        empty = self.__check()
        if empty == True:
            return
            
        print("1 to Combine")
        print("2 to Split")
        choice = input("Enter choice: ")
        
        if choice == "1":
            text = input("Enter numbers to combine: ")
            words = text.split()
            numbers = []
            for word in words:
                numbers.append(int(word))
            second = numpy.array(numbers).reshape(self.data.shape)
            answer = numpy.concatenate((self.data, second))
            print(answer)
        if choice == "2":
            parts = int(input("Enter how many parts: "))
            answer = numpy.array_split(self.data, parts)
            print(answer)

    def search_filter(self):
        """
        Method to search sort filter.
        Arguments taken: self.
        Return value: none.
        """
        empty = self.__check()
        if empty == True:
            return
            
        print("1 to Search")
        print("2 to Sort")
        print("3 to Filter")
        choice = input("Enter choice: ")
        
        if choice == "1":
            target = int(input("Enter value: "))
            answer = numpy.where(self.data == target)
            print(answer)
        if choice == "2":
            answer = numpy.sort(self.data)
            print(answer)
        if choice == "3":
            limit = int(input("Enter minimum: "))
            answer = self.data[self.data > limit]
            print(answer)

    def statistics(self):
        """
        Method to find statistics.
        Arguments taken: self.
        Return value: none.
        """
        empty = self.__check()
        if empty == True:
            return
            
        print("1 Sum")
        print("2 Mean")
        print("3 Median")
        print("4 Standard Deviation")
        print("5 Variance")
        print("6 Minimum")
        print("7 Maximum")
        print("8 Percentile")
        print("9 Correlation")
        choice = input("Enter choice: ")
        
        if choice == "1":
            print(numpy.sum(self.data))
        if choice == "2":
            print(numpy.mean(self.data))
        if choice == "3":
            print(numpy.median(self.data))
        if choice == "4":
            print(numpy.std(self.data))
        if choice == "5":
            print(numpy.var(self.data))
        if choice == "6":
            print(numpy.min(self.data))
        if choice == "7":
            print(numpy.max(self.data))
        if choice == "8":
            percent = int(input("Enter percent: "))
            print(numpy.percentile(self.data, percent))
        if choice == "9":
            text = input("Enter second array numbers: ")
            words = text.split()
            numbers = []
            for word in words:
                numbers.append(int(word))
            second = numpy.array(numbers).reshape(self.data.shape)
            answer = numpy.corrcoef(self.data.flatten(), second.flatten())
            print(answer)

```

```python
student = DataAnalytics()
student.show()

while True:
    print("Welcome to the NumPy Analyzer")
    print("Choose an option:")
    print("1 Create Array")
    print("2 Math Operations")
    print("3 Combine or Split")
    print("4 Search Sort Filter")
    print("5 Statistics")
    print("6 Exit")
    
    main = input("Enter your choice: ")
    
    if main == "1":
        student.create()
    if main == "2":
        student.maths()
    if main == "3":
        student.split_combine()
    if main == "4":
        student.search_filter()
    if main == "5":
        student.statistics()
    if main == "6":
        print(DataAnalytics.__doc__)
        print(student.__init__.__doc__)
        print(student._DataAnalytics__check.__doc__)
        print(DataAnalytics.show.__doc__)
        print(DataAnalytics.hello.__doc__)
        print(student.create.__doc__)
        print(student.maths.__doc__)
        print(student.split_combine.__doc__)
        print(student.search_filter.__doc__)
        print(student.statistics.__doc__)
        
        print("Goodbye")
        break
```
